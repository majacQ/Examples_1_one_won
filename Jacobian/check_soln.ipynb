Example of a non-trivial invertible polynomial map.

```python

import numpy as np
import sympy as sp
```

```python
z_1, z_2, z_3 = sp.symbols('z_1, z_2, z_3')
```

```python
g1, g2, g3 = (z_1 + 5*z_2**2 - 5*z_3, z_2 + z_3**2, z_1/5 + z_2**2)
```

```python
g1
```

```python
g2
```

```python
g3
```

```python
y_1, y_2, y_3 = sp.symbols('y_1, y_2, y_3')
eqns = [
    g1 - y_1, g2 - y_2, g3 - y_3
]
soln = sp.solve(eqns, (z_1, z_2, z_3))
assert len(soln) == 1
soln = soln[0]
```

```python
for s in soln:
    display(s)
```

```python
g1.subs({z_1: soln[0], z_2: soln[1], z_3: soln[2]}).expand()
```

```python
g2.subs({z_1: soln[0], z_2: soln[1], z_3: soln[2]}).expand()
```

```python
g3.subs({z_1: soln[0], z_2: soln[1], z_3: soln[2]}).expand()
```

```python
soln[0].subs({y_1: g1, y_2: g2, y_3: g3}).expand()
```

```python
soln[1].subs({y_1: g1, y_2: g2, y_3: g3}).expand()
```

```python
soln[2].subs({y_1: g1, y_2: g2, y_3: g3}).expand()
```
