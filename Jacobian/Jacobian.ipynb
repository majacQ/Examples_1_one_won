Checking Levent Alpöge's counter-example to the [Jacobian conjecture](https://en.wikipedia.org/wiki/Jacobian_conjecture) (as mentioned here: 

https://terrytao.wordpress.com/2026/07/21/a-digestion-of-the-jacobian-conjecture-counterexample/
) using sympy.

```python
import numpy as np
import sympy as sp
rng = np.random.default_rng(2026)
from Jacobian_fns import map_f, compute_detJ, find_repeated_roots
```

```python
z_1, z_2, z_3 = sp.symbols('z_1, z_2, z_3')
variables = (z_1, z_2, z_3)
```

```python

f1 = (1 + z_1 * z_2)**3 * z_3 + z_2**2 * (1 + z_1 * z_2) * (4 + 3* z_1 * z_2)
f2 = z_2 + 3 * z_1 * (1 + z_1 * z_2)**2 * z_3 + 3 * z_1 * z_2**2 * (4 + 3 * z_1* z_2)
f3 =  2 * z_1 - 3 * z_1**2 * z_2 - z_1**3 * z_3
```

```python
F = (f1, f2, f3)
```

```python
p1 = (0, 0, sp.Rational(-1, 4))
p2 = (1, sp.Rational(-3, 2), sp.Rational(13, 2))
p3 = (-1, sp.Rational(3, 2), sp.Rational(13, 2))
```

```python
map_f(fn=F, point=p1, variables=variables)
```

```python
map_f(fn=F, point=p2, variables=variables)
```

```python
map_f(fn=F, point=p3, variables=variables)
```

```python


detJ = compute_detJ(F, variables=variables)

print(f"\ndet(J) = {detJ}")

```

```python
num_bound = 8
den_bound = 4
find_repeated_roots(fn=F, num_bound=num_bound, den_bound=den_bound, variables=variables)
```
